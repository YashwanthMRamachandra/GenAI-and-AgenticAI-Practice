### Define `get_weather` tool

```python
def get_weather(city: str) -> dict:
  """
  Args:
      city (str): The name of the city (e.g., "New York", "London", "Tokyo").

  Returns:
      dict: A dictionary containing the weather information.
            Includes a 'status' key ('success' or 'error').
            If 'success', includes a 'report' key with weather details.
            If 'error', includes an 'error_message' key.
  """
  print(f"--- Tool: get_weather called for city: {city} ---") # Log tool execution
  city_normalized = city.lower().replace(" ", "") # Basic Normalization

  # Mock weather data
  mock_weather_db = {
      "newyork": {"status": "success", "report": "The weather in New York is sunny with a temperature of 25°C."},
      "london": {"status": "success", "report": "It's cloudy in London with a temperature of 15°C."},
      "tokyo": {"status": "success", "report": "Tokyo is experiencing light rain and a temperature of 18°C."},
  }

  if city_normalized in mock_weather_db:
    return mock_weather_db[city_normalized]
  else:
    return {"status": "error", "error_message": f"Sorry, I don't have weather information for '{city}'."}

print(get_weather("New York"))
print(get_weather("Paris"))
```

### Weather Agent

```python
# Use one of the model constants defined earlier
AGENT_MODEL = MODEL_GEMINI_2_5_FLASH # Starting with Gemini

weather_agent = Agent(
    model = AGENT_MODEL,
    name = "weather_agent",
    description = "Provides weather info for specified cities",
    instruction = "You are a helpful weather assistant."
                  "When the user asks for the weather in a specific city,"
                  "use the 'get_weather' tool to find the information."
                  "If the tool returns an error, inform the user politely"
                  "If the tool is successful, present the weather report clearly",
    tools=[get_weather],
)

print(f"Agent '{weather_agent.name}' created using model '{AGENT_MODEL}'.")
```

### Setup Runner

```python
# --- Session Management ---
# Key Concept: SessionService stores conversation history & state.
# InMemorySessionService is simple, non-persistent storage for this tutorial.
session_service = InMemorySessionService()

# Define constants for identifying the interaction context
APP_NAME = "weather_tutorial_app"
USER_ID = "weather_user_1"
SESSION_ID = "weather_session_001"

# Create the specific session where the conversation will happen
session = await session_service.create_session(
    app_name=APP_NAME,
    user_id=USER_ID,
    session_id=SESSION_ID,
)

print(f"Session created: App='{APP_NAME}', User='{USER_ID}', Session='{SESSION_ID}'")
```

```python
# --- Runner ---
# Key Concept: Runner orchestrates the agent execution loop.
runner = Runner(
    agent=weather_agent,
    app_name=APP_NAME,
    session_service=session_service
)

print(f"Runner created for agent '{runner.agent.name}'.")
```

### Interact with the Agent

```python
async def call_agent_async(query: str, runner, user_id, session_id):
  """Sends a query to the agent and prints the final response."""
  print(f"\n>>> User Query: {query}")

  # Prepare the user's message in ADK format
  content = types.Content(
      role = 'user',
      parts = [types.Part(text=query)]
  )

  final_response_text = "Agent did not produce a final response"

  # Key Concept: run_async executes the agent logic and yields Events.
  # We iterate through events to find the final answer
  async for event in runner.run_async(user_id=user_id, session_id=session_id, new_message=content):
    # You can uncomment the line below to see *all* events during execution
    # print(f"  [Event] Author: {event.author}, Type: {type(event).__name__}, Final: {event.is_final_response()}, Content: {event.content}")

    # Key Concept: is_final_response() marks the concluding message for the turn.
    if event.is_final_response():
      if event.content and event.content.parts:
        # Assuming text response in the first part
        final_response_text = event.content.parts[0].text
      elif event.actions and event.actions.escalate: # Handles potential errors/escalations
        final_response_text = f"Agent escalated: {event.error_message or 'No specific message.'}"
       # Add more checks here if needed (e.g., specific error codes)
      break # Stop processing events once the final response is found

  print(f"<<< Agent Response: {final_response_text}")
```

```python
# Load model directly
from transformers import AutoTokenizer, AutoModelForCausalLM
```

```python
tokenizer = AutoTokenizer.from_pretrained("google/gemma-3-270m")
model = AutoModelForCausalLM.from_pretrained("google/gemma-3-270m")
```

```python
model.eval()
```

### Poor prompt

```python
import torch

def summarize_text(text, max_new_tokens=120):
   prompt = f"""
Summarize the following text concisely:
{text}
Summary:
""".strip()
   inputs = tokenizer(
       prompt,
       return_tensors="pt",
       truncation=True,
       max_length=1024
   )
   with torch.no_grad():
       outputs = model.generate(
           **inputs,
           max_new_tokens=max_new_tokens,
           do_sample=False,           # deterministic
           temperature=0.7,
           top_p=0.9,
           repetition_penalty=1.1,
           eos_token_id=tokenizer.eos_token_id
       )
   generated_text = tokenizer.decode(
       outputs[0],
       skip_special_tokens=True
   )

   # Extract only the summary portion
   return generated_text.split("Summary:")[-1].strip()
```

```python
text = """
Large Language Models (LLMs) have transformed natural language processing by
enabling machines to generate, summarize, and reason over text. However,
challenges such as hallucinations, governance, cost, and scalability remain.
Organizations are increasingly adopting retrieval-augmented generation (RAG)
and agent-based architectures to mitigate these challenges.
"""
```

```python
summary = summarize_text(text)
print(summary)
```

### Prompt Engineer

```python
def summarize_text_enhanced(text, max_new_tokens=1000):
   prompt = f"""
You are an expert technical writer.
Summarize the following text in 2-3 sentences, preserving key technical details.

Text:
{text}

Concise Summary:
"""
   inputs = tokenizer(
       prompt,
       return_tensors="pt",
       truncation=True,
       max_length=1024
   )
   with torch.no_grad():
       outputs = model.generate(
           **inputs,
           max_new_tokens=max_new_tokens,
           do_sample=False,           # deterministic
           temperature=0.7,
           top_p=0.9,
           repetition_penalty=1.1,
           eos_token_id=tokenizer.eos_token_id
       )
   generated_text = tokenizer.decode(
       outputs[0],
       skip_special_tokens=True
   )

   # Extract only the summary portion
   return generated_text.split("Summary:")[-1].strip()
```

```python
summary = summarize_text_enhanced(text)
print(summary)
```

### Extraction process

```python
def summarize_text_extract(text, max_new_tokens=500):
   prompt = f"""
You MUST summarize the text below.
DO NOT define terms.
DO NOT add new information.
ONLY rewrite the content in shorter form.
TEXT:
{text}
SUMMARY (rewrite only):
"""
   inputs = tokenizer(
       prompt,
       return_tensors="pt",
       truncation=True,
       max_length=1024
   )
   with torch.no_grad():
       outputs = model.generate(
           **inputs,
           max_new_tokens=max_new_tokens,
           do_sample=False,           # deterministic
           temperature=0.7,
           top_p=0.9,
           repetition_penalty=1.1,
           eos_token_id=tokenizer.eos_token_id
       )
   generated_text = tokenizer.decode(
       outputs[0],
       skip_special_tokens=True
   )

   # Extract only the summary portion
   return generated_text.split("Summary:")[-1].strip()
```

```python
summary = summarize_text_extract(text)
print(summary)
```

## Evaluation Metrics

```python
# !pip install evaluate rouge-score nltk sacrebleu bert-score
```

```python
import evaluate

rouge = evaluate.load("rouge")

def rouge_scores(predictions, references):
   """
   predictions: List[str] # Model summary
   references: List[str]  # Human reference
   """
   results = rouge.compute(
       predictions=predictions,
       references=references
   )
   return results
```

```python
prediction = summary
reference = """
Large Language Models enable text generation and reasoning but face challenges
such as hallucinations and cost. RAG and agent-based systems help mitigate
these challenges.
"""
```

```python
import pprint

scores = rouge_scores([prediction], [reference])
pprint.pprint(scores, indent=4, sort_dicts=False, width=80)
```

```python
import sacrebleu

def bleu_score(prediction, reference):
   bleu = sacrebleu.sentence_bleu(prediction, [reference])

   return bleu.score
```

```python
scores = bleu_score(prediction, reference)
print(scores)
```

```python
from bert_score import score

def bertscore(predictions, references):
   P, R, F1 = score(
       predictions,
       references,
       lang="en",
       verbose=False
   )
   return {
       "precision": P.mean().item(),
       "recall": R.mean().item(),
       "f1": F1.mean().item()
   }
```

```python
print(bertscore([prediction], [reference]))
```

```python
def compression_ratio(original, summary):
   return len(summary.split()) / len(original.split())
```

```python
print(compression_ratio(text, summary))
```

```python
def coverage(original, summary):
   original_tokens = set(original.lower().split())
   summary_tokens = summary.lower().split()
   overlap = sum(1 for t in summary_tokens if t in original_tokens)

   return overlap / max(1, len(summary_tokens))
```

```python
print(coverage(text, summary))
```

```python
def hallucination_rate(original, summary):
   original_tokens = set(original.lower().split())
   summary_tokens = set(summary.lower().split())
   hallucinated = summary_tokens - original_tokens

   return len(hallucinated) / max(1, len(summary_tokens))
```

```python
print(hallucination_rate(text, summary))
```

```python
def evaluate_summary(original, prediction, reference):
   return {
       "rouge": rouge_scores([prediction], [reference]),
       "bleu": bleu_score(prediction, reference),
       "bert_score": bertscore([prediction], [reference]),
       "compression_ratio": compression_ratio(original, prediction),
       "coverage": coverage(original, prediction),
       "hallucination_rate": hallucination_rate(original, prediction),
   }
```

```python
metrics = evaluate_summary(text, summary, reference)
for k,v in metrics.items():
  print(k,": ", v)
```
